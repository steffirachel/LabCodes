### Cell 1: Import libraries and create data

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Condition": ["Normal", "Congestion", "Failure"],
    "Packets Sent": [10253, 29531, 10253],
    "Packets Received": [6671, 6199, 6668],
    "PDR (%)": [65.0639, 20.9915, 65.0346],
    "Average Delay (ms)": [598.681, 2235.14, 595.095],
    "Routing Table Changes": [1, 2, 5]
}

df = pd.DataFrame(data)
```

### Cell 2: Display the data

```python
print(df)
```

### Cell 3: Graph for Packet Delivery Ratio

```python
plt.figure(figsize=(7, 4))
plt.bar(df["Condition"], df["PDR (%)"])

plt.title("RIP - Packet Delivery Ratio")
plt.xlabel("Network Condition")
plt.ylabel("PDR (%)")
plt.ylim(0, 100)

plt.show()
```

**Inference:** The PDR is lower during congestion, while the normal and failure conditions have similar PDR values.

### Cell 4: Graph for Average Delay

```python
plt.figure(figsize=(7, 4))
plt.bar(df["Condition"], df["Average Delay (ms)"])

plt.title("RIP - Average Delay")
plt.xlabel("Network Condition")
plt.ylabel("Average Delay (ms)")

plt.show()
```

**Inference:** Congestion significantly increases average packet delay.

### Cell 5: Graph for Packets Sent and Received

```python
plt.figure(figsize=(7, 4))

plt.bar(df["Condition"], df["Packets Sent"], label="Packets Sent")
plt.bar(df["Condition"], df["Packets Received"], label="Packets Received")

plt.title("RIP - Packets Sent vs Received")
plt.xlabel("Network Condition")
plt.ylabel("Number of Packets")
plt.legend()

plt.show()
```

**Inference:** During congestion, more packets are sent, but the number received does not increase proportionally.

### Cell 6: Graph for Routing Table Changes

```python
plt.figure(figsize=(7, 4))
plt.bar(df["Condition"], df["Routing Table Changes"])

plt.title("RIP - Routing Table Changes")
plt.xlabel("Network Condition")
plt.ylabel("Observed Snapshot Changes")

plt.show()
```

**Inference:** The failure scenario has the highest number of observed routing-table snapshot changes. This is not a direct count of routing-update messages.

### Cell 7: Final conclusion

```python
print("CONCLUSION")
print("-----------")
print("1. Congestion reduced the packet delivery ratio.")
print("2. Congestion increased the average packet delay.")
print("3. The failure scenario showed more routing-table snapshot changes.")
print("4. Normal and failure scenarios had similar PDR and delay values.")
print("5. More results are needed to compare RIP with AODV and Global Routing.")
```
